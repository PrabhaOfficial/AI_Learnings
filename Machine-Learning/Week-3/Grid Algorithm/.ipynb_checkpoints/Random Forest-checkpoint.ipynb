```python
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True,dtype=int) # convert categorical into numerical dataset so that a model can be created
```

```python
dataset
```

```python
dataset.columns # get the column headers only
```

```python
#Get the independent/Input dataset alone
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']] 
```

```python
independent # Display the independant dataset
```

```python
#Get the dependent/output dataset alone
dependent=dataset[["charges"]] 
```

```python
# split the dataset into test set and training set
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(independent, dependent, test_size=0.30,random_state=0)
```

```python
# Train the model using the training dataset
import  sklearn.model_selection as GSCV
from sklearn.ensemble import RandomForestRegressor
param_grid = {'criterion' :['squared_error', 'absolute_error','poisson'],'max_features': ['sqrt', 'log2'],'bootstrap':[True,False],'min_samples_split':[2,5,10], 'min_samples_leaf':[2,5,10],'max_depth':[5,10]}

grid =  GSCV.GridSearchCV(RandomForestRegressor(),param_grid,refit=True,verbose=3,n_jobs=-1)
grid.fit(X_train,y_train)
re = grid.cv_results_
pd.DataFrame.from_dict(re)

```

```python
#testdata run againt the pediction. 
grid_pred=grid.predict(X_test)
#from sklearn.metrics import r2_score
#r_score=r2_score(y_test,grid_pred)
print("The r2 score value for best parameter {}:".format(grid.best_params_),grid.best_score_)
pred=grid.predict([[32,43.900,2,0,1]])
print(pred) #4449.46200
```

```python
import pickle
filename="finalized_model_RandomForest.sav"
pickle.dump(grid,open(filename,'wb'))
```
