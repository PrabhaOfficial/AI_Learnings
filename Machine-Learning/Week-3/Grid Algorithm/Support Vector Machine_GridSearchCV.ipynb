```python
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True,dtype=int) # convert categorical into numerical dataset so that a model can be created
```

```python
dataset
```

```python
dataset.columns # get the column headers only
```

```python
#Get the independent/Input dataset alone
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']] 
```

```python
independent # Display the independant dataset
```

```python
#Get the dependent/output dataset alone
dependent=dataset[["charges"]] 
```

```python
# Train the model using the training dataset
import  sklearn.model_selection as GSCV
from sklearn.svm import SVR
param_grid = {'kernel':['linear', 'rbf', 'poly', 'sigmoid'],'C':[10,1,.1,5000,4000],'gamma': ['auto', 'scale']}
```

```python
grid =  GSCV.GridSearchCV(SVR(),param_grid,refit=True,verbose=3,n_jobs=-1)
grid.fit(independent,dependent)
re = grid.cv_results_
pd.DataFrame.from_dict(re)
```

```python
print("The r2 score value for best parameter {}:".format(grid.best_params_),grid.best_score_)

pred=grid.predict([[32,43.900,2,0,1]])
print(pred) #4449.46200
```
