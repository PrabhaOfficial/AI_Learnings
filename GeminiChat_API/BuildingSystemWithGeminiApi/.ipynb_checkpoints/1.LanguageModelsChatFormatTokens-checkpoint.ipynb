```python
from google import genai
from google.genai import types
```

```python
client = genai.Client(api_key="APIKEy")
chat = client.chats.create(
    model="gemini-2.5-flash")

```

## This is an example of having temperature , max_output_tokens, top_k,top_k

#### temperature: Controls randomness/creativity of the response.

Low (0.0 - 0.3) → more deterministic and focused.
High (0.7 - 1.5) → more creative and varied.



#### max_output_tokens: Maximum number of tokens the model can generate in its response.

Higher value → longer responses.
Lower value → shorter responses.



#### top_p (Nucleus Sampling): Limits token selection to the smallest set of tokens whose cumulative probability reaches p.

Lower value → safer, more focused responses.
Higher value → more diverse responses.



#### top_k: Limits token selection to the k most likely next tokens.

Lower value → more predictable responses.
Higher value → more varied responses.

Token details
-----------
#### prompt_token_count: 
Number of input tokens sent to the model.

#### candidates_token_count:
Number of output tokens generated by the model.

#### thoughts_token_count: 
Number of internal reasoning/thinking tokens used by the model during generation (available for models that support thinking).

#### total_token_count: 
Total tokens consumed, including prompt tokens, output tokens, thought tokens, and any other model-specific token usage.



```python
def CustomChatBot(topp,topk,temp,maxToken):
    response= chat.send_message("write me a story about a happy carrot",
                                config= types.GenerateContentConfig(
                                    temperature=temp,
                                    max_output_tokens=maxToken,                                 
                                    top_p=topp,
                                    top_k=topk,
                                    # multi line
                                    systemInstruction="""You are an assistant who responds in the style of Dr Seuss.
                                    All your responses must be one sentence long.
                                    """
                                    ))
    return response.text, response.usage_metadata

response, usage = CustomChatBot(0.8,1000,0.7,1000)
print(f"The prompt token usage {usage.prompt_token_count}")
print(f"The completion token usage {usage.candidates_token_count}")
print(f"The thoughts token usage {usage.thoughts_token_count}") 
print(f"The total token usage {usage.total_token_count}")
print(f"The response text {response}")
```

This is a good example of why LLMs are not guaranteed to perform character-level operations accurately. They predict likely text rather than actually executing a deterministic string-reversal algorithm.
For tasks like:

*Reversing strings

*Counting letters exactly

*Counting words exactly

*Arithmetic

LLMs can make mistakes, especially at higher temperatures.
