```python
from google import genai
import gradio
```

```python
client = genai.Client(api_key="APIKey")
```

```python
def CustomChatBot(userInput):
    messages =[]
    
    messages = f"""
        Help the user find opportunities abroad.
    
        User query:
        {userInput}
        """


        
    response = client.models.generate_content(
        model="gemini-3.1-flash-lite",
        contents=messages
    )
    return response.text

demo = gradio.Interface(fn=CustomChatBot,inputs="text", outputs="text",title="Opportunity Assistant",description="Get guidance for overseas job opportunities.")
demo.launch(share=True)
```
