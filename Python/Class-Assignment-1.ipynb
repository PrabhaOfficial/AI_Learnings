```python
class SubfieldsInAI():
    def Subfields(subFds):
        print("Sub-fields in AI are:")
        for field in subFds:
            print(field)

subFds = ["Machine Learning","Neural Networks","Vision","Robotics","Speech Processing","Natural Language Processing"]
SubfieldsInAI.Subfields(subFds)
```

```python
class OddEven():
    def oddEven(num):
        if (num%2 == 0):
            print(num,"is a Even number")
        else:
            print(num,"is a Odd number")

num = int(input("Enter a num: "))
OddEven.oddEven(num)
```

```python
class ElegiblityForMarriage():
    def Eligible(age,gender):
        if(gender == "male"):        
            if (age < 21):
                return ("Not Eligible")
            else:
                return ("Eligible")
        elif(gender == "female"):
            if (age < 18):
                return ("Not Eligible")
            else:
                return ("Eligible")

age = int(input("Your Age: "))
gender = input("Your Gender: ")
print(ElegiblityForMarriage.Eligible(age,gender))
```

```python
class FindPercent():
    def percentage():
        sub1 = int(input("Subject1= "))
        sub2 = int(input("Subject2= "))
        sub3 = int(input("Subject3= "))
        sub4 = int(input("Subject4= "))
        sub5 = int(input("Subject5= "))
        total = sub1+sub2+sub3+sub4+sub5
        per=total/5
        print ("Total: " , total)
        print("Percentage: ", per)

FindPercent.percentage()
```
