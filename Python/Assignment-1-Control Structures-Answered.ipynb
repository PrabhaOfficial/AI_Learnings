```python
height= float(input("Enter the height in meter ="))
weight= float(input("Enter the weight in kg ="))
```

```python
bmi = weight/(height ** 2)
print("The BMI index is",bmi)
```

```python
if (bmi < 18.5):
    print("underweight")
elif (18.5 <= bmi <= 24.9):
    print("normal")
elif (25 <= bmi <= 29.9):
    print("overweight")
elif (bmi > 30):
    print("very weight")
```
