## if-else

```python
age=int(input("Enter the age"))
```

```python
if(age<=18):
    print("Children")
else:
    print("Adult")
```

## if-elif

```python
age=int(input("Enter the age"))
```

```python
if(age<18):
    print("Children")
elif(age<35):
    print("Adult")
elif(age<59):
    print("Citizen")
else:
    print("Senior Citizen")
```

## For loop

```python
ageList = [34,56,78,98,12]
ageList
```

```python
for age in ageList:
    if(age<18):
        print("Children")
    elif(age<35):
        print("Adult")
    elif(age<59):
        print("Citizen")
    else:
        print("Senior Citizen")
```

### Odd number
