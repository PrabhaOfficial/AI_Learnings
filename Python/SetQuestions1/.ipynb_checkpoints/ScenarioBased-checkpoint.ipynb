```python
#2.	Scenario: A program processes a list of numbers and needs to find the largest value.
def FindLargest():
    numList = [-34,-56,-78,-98,-12]
    largestnum=numList[0]
    for num in numList:
        if(num >largestnum):
            largestnum= num
    return largestnum
print(FindLargest())
```

```python
#3.	Scenario: A company provides employees with a 10% bonus if their salary exceeds $50,000.
def CalcBonus():
    sal=int(input("Enter the salary $: "))
    bonus=0
    if (sal>50000):
        bonus=sal*.1
        print("The bonus amount is ", bonus)
    else:
        print("No bonus")

CalcBonus()
```

```python
#6.	Scenario: A grading system determines whether a student has passed or failed based on their score.

def PassOrFail():
    subjects=["English","Maths","Physics","Computer","Chemistry"]
    subMarks={}
    #subMarks=int(input(f"Enter mark {n+1}: ")) for n in range(5)
    for sub in subjects:
        subMarks[sub]=int(input(f"Enter the {sub} marks:"))
    
    for mark in subMarks.values():
       if(mark<40):
           print("Failed")
           return
    print("Pass")
    
PassOrFail()
```

```python
#9.Scenario: A calendar system verifies whether a given year is a leap year based on standard leap year rules.
#Write logic to determine whether a given year is a leap year

year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```
